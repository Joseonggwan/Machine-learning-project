# 학생 성적 예측 및 영향 요인 분석

## 1. 프로젝트 개요

학생의 학업 성취도는 개인의 학습 습관뿐만 아니라 가정환경, 생활습관, 학교생활 등 다양한 요인에 의해 영향을 받는다. 본 프로젝트의 목적은 학생들의 최종 성적(G3)에 영향을 미치는 주요 요인을 분석하고, 머신러닝 기법을 활용하여 학생 성적을 예측하는 것이다.

기존 분석에서는 10명의 학생 데이터를 활용하여 단순 선형회귀 모델을 적용하였다. 그러나 데이터 규모가 매우 작고 변수의 다양성이 부족하여 일반화 성능에 한계가 존재하였다.

따라서 본 프로젝트에서는 실제 학생 성적 데이터셋(Student Performance Dataset)을 활용하여 보다 신뢰성 있는 분석을 수행하였으며, 다양한 머신러닝 모델 비교와 통계적 검정을 통해 학생 성적에 영향을 주는 핵심 요인을 도출하였다.

# 4. 탐색적 데이터 분석 (EDA)

## 4.1 최종 성적 분포

최종 성적(G3)의 분포를 확인한 결과 학생들의 성적이 넓게 분포되어 있었으며 일부 학생은 매우 낮은 성적을 기록하였다.

이는 학생 간 학업 성취도 차이가 상당히 존재함을 의미한다.

## 4.2 상관관계 분석

상관관계 분석 결과 다음 변수들이 성적과 비교적 높은 관련성을 보였다.

### 양의 상관관계

* studytime
* Fedu

### 음의 상관관계

* failures
* absences
* age

특히 낙제 경험(failures)은 최종 성적과 가장 강한 음의 상관관계를 나타냈다.

## 4.3 주요 변수 시각화

EDA를 통해 다음과 같은 경향을 확인하였다.

* 공부 시간이 많을수록 성적이 높아지는 경향
* 결석 횟수가 많을수록 성적이 낮아지는 경향
* 과거 낙제 경험이 많을수록 성적이 크게 감소하는 경향
* 부모 교육 수준이 높을수록 성적이 향상되는 경향

# 6. 모델 성능 비교

모델 성능 비교 결과 Random Forest Regressor가 가장 우수한 예측 성능을 보였다.

### Linear Regression

* 변수 간 선형 관계만 반영 가능
* 해석은 용이하지만 복잡한 패턴 학습에는 한계 존재

### KNN Regressor

* 국소적 패턴 학습 가능
* 데이터 분포에 민감

### Random Forest Regressor

* 비선형 관계 학습 가능
* 과적합에 강함
* 가장 높은 예측 정확도 달성

따라서 최종 모델로 Random Forest를 선정하였다.

# 10. 결론

본 프로젝트에서는 학생 성적 예측과 영향 요인 분석을 수행하였다.

분석 결과 다음과 같은 사실을 확인할 수 있었다.

* 결석 횟수는 성적에 가장 큰 영향을 미치는 요인이다.
* 과거 낙제 경험은 미래 학업 성취도에 부정적인 영향을 준다.

Random Forest 모델은 학생 성적 예측에서 가장 우수한 성능을 보였으며, 학생 성적에 영향을 미치는 주요 요인을 효과적으로 설명할 수 있었다.
