```python
# 📊 학생 생활 요인이 학업 성적에 미치는 영향 분석

## 1. 프로젝트 개요

본 프로젝트는 학생의 다양한 생활 요인(개인 정보, 학습 습관, 사회적 활동 등)이
학업 성취도에 미치는 영향을 분석하는 것을 목표로 한다.

특히 단일 요인이 아닌 여러 요인을 종합적으로 분석하여
성적에 가장 큰 영향을 미치는 핵심 요소를 도출하고자 한다.

---

## 2. 데이터셋 분석

### 2.1 데이터 출처
- Kaggle: Student Alcohol Consumption Dataset
- 원본: UCI Machine Learning Repository

- [Student Alcohol Consumption Dataset](https://www.kaggle.com/datasets/uciml/student-alcohol-consumption)
---

### 2.2 데이터 구성

- 총 데이터 수: 약 649개
- 변수 수: 약 30개 이상
- 데이터 유형: 설문 기반 + 학업 기록 데이터

---

### 2.3 사용 변수

#### 🎯 종속 변수 (Target)
- G3: 최종 성적

---

#### 🔑 독립 변수 (Features)

### 📌 개인 및 배경 정보
- school: 학교
- sex: 성별
- age: 나이
- reason: 학교 선택 이유

---

### 📌 학습 관련 요인
- traveltime: 통학 시간
- studytime: 주간 학습 시간
- failures: 과거 낙제 횟수
- schoolsup: 추가 교육 지원 여부
- paid: 유료 수업 여부

---

### 📌 활동 및 환경 요인
- activities: 방과후 활동 여부
- nursery: 유치원 교육 여부
- higher: 고등교육 진학 희망 여부
- internet: 인터넷 사용 가능 여부

---

### 📌 사회적 및 생활 요인
- romantic: 연애 여부
- famrel: 가족 관계의 질
- freetime: 방과후 여가 시간
- goout: 외출 빈도

---

### 📌 건강 및 행동 요인
- Dalc: 평일 음주량
- Walc: 주말 음주량
- health: 건강 상태
- absences: 결석 횟수

---

### 2.4 데이터 전처리

- 결측치 확인 및 처리 (결측치 거의 없음)
- 범주형 변수 인코딩 (Label Encoding / One-Hot Encoding)
- 파생 변수 생성:
  - total_alcohol = Dalc + Walc
- 필요 시 데이터 정규화/표준화 수행

---

## 3. 모델링 분석

### 3.1 사용 알고리즘
- Linear Regression
- Random Forest Regression

---

### 3.2 모델 선택 이유

- Linear Regression:
  - 변수 간 관계를 해석하기 용이함
- Random Forest:
  - 비선형 관계를 효과적으로 학습 가능
  - 변수 중요도 분석 가능
  - 과적합 방지에 효과적

---

## 4. 성능 평가 분석

### 4.1 평가 지표
- RMSE (Root Mean Squared Error)
- R² Score

---

### 4.2 지표 설명

- RMSE:
  - 예측값과 실제값 간의 평균 오차를 나타냄
- R² Score:
  - 모델이 데이터를 설명하는 비율

---

### 4.3 성능 개선 방안

- 하이퍼파라미터 튜닝 (Grid Search)
- 변수 중요도 기반 Feature Selection
- 파생 변수 추가 생성

---

## 5. 분석 목표 및 핵심 질문

본 프로젝트는 다음과 같은 질문에 답하는 것을 목표로 한다.

1. 어떤 생활 요인이 성적에 가장 큰 영향을 미치는가?
2. 학습 시간과 결석 횟수 중 어떤 요소가 더 중요한가?
3. 사회적 활동(외출, 연애 등)은 성적에 어떤 영향을 미치는가?
4. 음주와 건강 상태는 학업 성취도와 어떤 관계가 있는가?

---

## 6. 기대 효과

본 프로젝트를 통해 학생의 다양한 생활 요인과 학업 성취도 간의 관계를 분석하고,
단순 예측을 넘어 실제 생활 개선에 활용 가능한 인사이트를 도출할 수 있을 것으로 기대된다.

또한 데이터 기반 분석을 통해 학생 관리 및 교육 방향 설정에 기여할 수 있다.

---

## 7. 결론 (예상)

본 연구는 다양한 생활 요인을 종합적으로 분석하여
성적에 가장 큰 영향을 미치는 핵심 요인을 도출하는 것을 목표로 한다.

특히 학습 시간과 결석 횟수와 같은 요소가
성적에 중요한 영향을 미칠 것으로 예상된다.

---

## 📌 핵심 요약

> 학생의 성적은 개인적, 학습적, 사회적 요인이 복합적으로 작용한 결과이다.
```
